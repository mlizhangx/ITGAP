```python
mc_pas_human_peptide = mc_pas_human[['CDR3.beta.aa', 'Pathology','Epitope.peptide']]
mc_pas_human_peptide = mc_pas_human_peptide[~mc_pas_human_peptide['Epitope.peptide'].isnull()]
mc_pas_human_peptide = mc_pas_human_peptide[~(mc_pas_human_peptide['CDR3.beta.aa'].str.contains('\*') ==True)]
mc_pas_human_peptide.columns = ['CDR3','Epitope_type','Epitope']
mc_pas_human_peptide
```

```python
vdjdb = pd.read_csv('./VDJ_db_human_033026.tsv', sep='\t')
vdjdb = vdjdb[vdjdb['Epitope species'] == "HomoSapiens"]
vdjdb = vdjdb[vdjdb['Gene'] == 'TRB']
```

```python
vdjdb.columns
```

```python
vdjdb_clean = vdjdb[['CDR3','Epitope', 'Epitope gene']].copy()
vdjdb_clean.columns = ['CDR3','Epitope','Epitope_type']
vdjdb_clean
```

```python
IEDB_tumor_epitope = pd.read_csv('./Apexigen_data/IEDB_tumor_epitope.csv', index_col=0)
IEDB_tumor_epitope.rename(columns={"Epitope_Name":"peptide"}, inplace=True)
IEDB_tumor_epitope = IEDB_tumor_epitope.drop_duplicates(subset=['peptide'])
IEDB_tumor_epitope
```
